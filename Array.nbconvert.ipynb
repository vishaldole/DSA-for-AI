## Arrays

###1.Array

```python
import array
```

```python
# creation

arr1 = array.array('i', [10, 20, 30, 40, 50])
print(arr1)
```

```python
# access

arr1[2]
```

```python
arr1[3]
```

```python
#traverse

for element in arr1:
    print(element, end =" ")
```

```python
# insertion 

arr1.insert(3, 35);
```

```python
print(arr1)
```

```python
# Deletion

arr1.remove(30) #remove first occerance of 30
print(arr1)
```

```python
#Updation

arr1[2] = 30

print(arr1)
```

```python
#Sorting

arr2 = array.array('i', [60, 70, 80])

arr3 = array.array('i', [4, 7, 3, 9, 2, 6])

print(f'arr2 is {arr2}')

print(f'arr3 is {arr3}')
```

```python
sorted(arr3)
```

```python
#convert this sorted list to sorted array
sorted_arr3 = array.array('i', sorted(arr3))
print(sorted_arr3)
```

```python
#Merge

print(f'arr1 is {arr1}')

print(f'arr2 is {arr2}')

print(f'mering arr1 and arr2, we get {arr1 + arr2}')
```

```python
arr4 = arr1 + arr2

arr4
```

```python
# Splitting

ix = int(len(arr4) / 2 )
ix
```

```python
# splitting on ix index
arr4[:ix]
```

```python
arr4[ix:]
```

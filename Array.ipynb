## Arrays

### 1.Array

```python
import array
```

```python
# creation

arr1 = array.array('i', [10, 20, 30, 40, 50])
print(arr1)
```

```python
# access

arr1[2]
```

```python
arr1[3]
```

```python
#traverse

for element in arr1:
    print(element, end =" ")
```

```python
# insertion 

arr1.insert(3, 35);
```

```python
print(arr1)
```

```python
# Deletion

arr1.remove(30) #remove first occerance of 30
print(arr1)
```

```python
#Updation

arr1[2] = 30

print(arr1)
```

```python
#Sorting

arr2 = array.array('i', [60, 70, 80])

arr3 = array.array('i', [4, 7, 3, 9, 2, 6])

print(f'arr2 is {arr2}')

print(f'arr3 is {arr3}')
```

```python
sorted(arr3)
```

```python
#convert this sorted list to sorted array
sorted_arr3 = array.array('i', sorted(arr3))
print(sorted_arr3)
```

```python
#Merge

print(f'arr1 is {arr1}')

print(f'arr2 is {arr2}')

print(f'mering arr1 and arr2, we get {arr1 + arr2}')
```

```python
arr4 = arr1 + arr2

arr4
```

```python
# Splitting

ix = int(len(arr4) / 2 )
ix
```

```python
# splitting on ix index
arr4[:ix]
```

```python
arr4[ix:]
```

#### 2. Numpy

```python
import numpy as np
```

```python
# creation

np1 = np.array([10, 20, 30, 40, 50])
np2 = np.array([50, 60, 70])

print(np1)
print(np2)
```

```python
np1
```

```python
# access

np1[2]
```

```python
#traverse

for ele in np1:
    print(ele, end = " ")
```

```python
# insertion

np.insert(np1, 3, 35)
```

```python
# deletion

np.delete(np1, 3)
```

```python
#updation

np1[2] = 35
print(np1)
```

```python
# sorting

np3 = np.array([5, 3, 4, 2,6 , 9])

np3
```

```python
np.sort(np3)
```

```python
# merging

np3 = np.concatenate((np1, np2))

np3
```

```python
# splitting

split1, split2 = np.array_split(np3, 2)  ## 2 stands two parts
```
